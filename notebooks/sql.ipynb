Setup

```kotlin
USE {
    dependencies("org.postgresql:postgresql:42.7.8")
}
```

```kotlin
%use dataframe(1.0.0-Beta3n)
```

```kotlin
dataFrameConfig.version
```

```kotlin
val dbConfig = DbConnectionConfig("jdbc:postgresql://localhost:5432/test_db", "test_user", "test")
```

Reading tables

```kotlin
val customers = DataFrame.readSqlTable(dbConfig, "customers")

customers
```

```kotlin
val dfBrad = DataFrame.readSqlQuery(dbConfig, "SELECT * FROM customers WHERE firstname = 'Brad'")

dfBrad
```

```kotlin
val dataframes = DataFrame.readAllSqlTables(dbConfig)

dataframes
```

```kotlin
dataframes.keys
```

```kotlin
val dfBooks = dataframes["books"]

dfBooks
```

Get different schemas for the same tables

```kotlin
val ordersSchema = DataFrame.getSchemaForSqlTable(dbConfig, "orders")

ordersSchema
```

```kotlin
val dfOrders = DataFrame.readSqlTable(dbConfig, "orders")

dfOrders.schema()
```

```kotlin
val customersSchema = DataFrame.getSchemaForSqlTable(dbConfig, "customers")

customersSchema
```
